```python
import pandas as pd
```

```python
data = pd.read_csv('Housing.csv')
data.head()
```

```python
data.info()
```

```python
data.isna().sum()
```

```python
data.duplicated().sum()
```

```python
numdata = data.select_dtypes(include='number')
otherdata = data.select_dtypes(include='object')
numdata.head(),otherdata.head()
```

```python
x = numdata.drop('price',axis=1)
y = data['price']
x.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(x,y,test_size=0.30,random_state=42)
print("X train:", X_train.shape)
print("X test:", X_test.shape)
print("Y train:", Y_train.shape)
print("Y test:", Y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train,Y_train)
```

```python
lr.score(X_train,Y_train)
```

```python
y_train_pred = lr.predict(X_train)
y_test_pred = lr.predict(X_test)
```

```python
res = pd.DataFrame({
    "Actual": Y_test.values,
    "Predicted": y_test_pred
})
res.head(10)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
tmae = mean_absolute_error(Y_train, y_train_pred)
tmse = mean_squared_error(Y_train, y_train_pred)
trmse = tmse ** 0.5 
tr2 = r2_score(Y_train, y_train_pred)

print("Train Data Set performance")
print("Mean Absolute Error :", tmae)
print("Mean Squared Error  :", tmse)
print("TRMSE:", trmse)
print("R2  :", tr2)
```

```python
tmae = mean_absolute_error(Y_test, y_test_pred)
tmse = mean_squared_error(Y_test, y_test_pred)
trmse = tmse ** 0.5
tr2 = r2_score(Y_test,y_test_pred)

print("test Data Set ")
print("Mean Absi=olute Error :", tmae)
print("Mean Squatrd Error :", tmse)
print("TRMSE:", trmse)
print("R2  :", tr2)
```
