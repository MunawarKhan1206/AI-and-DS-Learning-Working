```python
import pandas as pd
```

```python
data = pd.read_csv('student_habits_performance.csv')
data.head()
```

```python
data.isna().sum()
```

```python
data.duplicated().sum()
```

```python
numdata = data.select_dtypes(include='number')
otherdata = data.select_dtypes(include='object')
```

```python
numdata
otherdata
```

```python
x = numdata.drop('exam_score',axis=1)
y = numdata['exam_score']
x.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(x,y,test_size=0.30,random_state=42)
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("Y_train:", Y_train.shape)
print("Y_test:", Y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train,Y_train)
```

```python
lr.score(X_train,Y_train)
```

```python
Y_train_pred = lr.predict(X_train)
Y_test_pred = lr.predict(X_test)
```

```python
res = pd.DataFrame(
   { 
    "Actual" : Y_test.values,
    "Predicted" : Y_test_pred
    }
)
```

```python
res
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
tmean = mean_absolute_error(Y_train, Y_train_pred)
tmse = mean_squared_error(Y_train, Y_train_pred)
trmse = tmse ** 0.5  
tr2 = r2_score(Y_train, Y_train_pred)
print("Ttraing Data Set ")
print("Mean Absi=olute Error :", tmean)
print("Mean Squatrd Error :", tmse)
print("TRMSE:", trmse)
print("R2  :", tr2)
```

```python
tmae = mean_absolute_error(Y_test, Y_test_pred)
tmse = mean_squared_error(Y_test, Y_test_pred)
trmse = tmse ** 0.5
tr2 = r2_score(Y_test, Y_test_pred)

print("Test Data Set ")
print("Mean Absi=olute Error :", tmean)
print("Mean Squatrd Error :", tmse)
print("TRMSE:", trmse)
print("R2  :", tr2)
```
